STANDARDIZING VARIABLE NAMES

```python
import re
```

```python
# we define a function to clean a variable name
def clean_variable_name(name):
    """
    Convert an original questionnaire
column name into
    a clean, analysis-friendly
snake_case variable name.
    """
    
    #convert to string
    name = str(name)
    
    #remove leading/trailing spaces
    name = name.strip()
    
    #convert to lowercase
    name = name.lower()
    
    #replace special characters with spaces
    name = re.sub(r'[^a-z0-9]+','_',name)
    
    #remove repeated underscores
    name = re.sub(r'_+', '_',name)
    
    #remove underscores at beginning/end
    name = name.strip('_')
    
    return name
```

```python
# Automatic Standardization
# parsing the function
business_clean = business.copy()
customer_clean = customer.copy()
business_clean.columns = [clean_variable_name(col) for col in business_clean.columns]
customer_clean.columns = [clean_variable_name(col) for col in customer_clean.columns]
```

DETERMINING THE MEASUREMENT LEVEL OF EVERY VARIABLE

```python
# Function to determine measurement level
def determine_measurement_level(series):
    # remove missing values
    x = series.dropna()
    # if there are no observations
    if len(x) == 0:
        return "Unknown"
    #Numeric variables
    
    if pd.api.types.is_numeric_dtype(x):
        unique_values = x.nunique()
        
        # Binary variable
        if unique_values == 2:
            return "Nominal (Binary)"
        
        # Integer/count variable
        if pd.api.types.is_integer_dtype(x):
            if x.min(x) >= 0:
                return "Ratio"
        #Decimal numeric variable
        if pd.api.types.is_float_dtype(x):
            return "Numeric - Review"
    
    # Text Variables
    
    x_string = x.astype(str).str.strip()
    unique_values = x_string.nunique()
    # convert to lowercase for checking
    x_lower = x_string.str.lower()
    
    # Binary categorical variables
    
    binary_patterns = [
        {"yes","no"},
        {"y","n"},
        {"agree","disagree"},
        {"male","female"},
        {"true","false"}
    ]
    
    observed_values = set(x_lower.unique())
    
    for pattern in binary_patterns:
        if observed_values.issubset(pattern) and len(observed_values) == 2:
            return "Nominal (Binary)"
        
    #Ordinal response patterns
    
    ordinal_keywords = [
        "very good",
        "good",
        "average",
        "poor",
        "strongly agree",
        "agree",
        "neutral",
        "disagree",
        "strongly disagree",
        "always",
        "often",
        "sometimes",
        "rarely",
        "never",
        "high",
        "medium",
        "low",
        "very high",
        "very low",
        "excellent",
        "fair",
        "bad",
        "better",
        "worse"
    ]
    
    #check whether values contain ordinal language
    
    if any(any (keyword in value for keyword in ordinal_keywords)
           for value in x_lower
    ):
        return "Ordinal"
    
    # Age-group patterns
    
    age_pattern = r"\d+\s*[--]\s*\d+|\d+\s*(and above|years|yrs)"
    
    if x_string.str.contains(
        age_pattern,
        case=False,
        regex=True,
        na=False
    ).mean() > 0.5:
        return "Ordinal (Age Group)"
    
    # Likert-style numeric categories stored as text
    
    numeric_text = pd.to_numeric(x_string, errors="coerce")
    if numeric_text.notna().mean() > 0.95:
        unique_numeric = numeric_text.nunique()
        
        # Typical rating scale
        if unique_numeric <= 11:
            return "Ordinal / Rating Scale"
        return "Ratio"
    
    # open ended text
    
    average_length = x_string.str.len().mean()
    if average_length > 30 or unique_values > 0.5 *len(x_string):
        return "Text / Open-ended"
    
    # Default categorical variable
    return "Nominal"

# CREATE MEASUREMENT-LEVEL DICTIONARY

def create_measurement_dictionary(df, dataset_name):
    results = []
    for column in df.columns:
        series = df[column]
        measurement_level = determine_measurement_level(series)
        results.append({
            "dataset": dataset_name,
            "variable": column,
            "data_type":str(series.dtype),
            "number_of_observations":series.isna().sum(),
            "number_missing":series.isna().sum(),
            "number_unique":series.nunique(dropna=True),
            "measurement_level":measurement_level
        })
    return pd.DataFrame(results)
```

```python
from scipy.stats import mannwhitneyu
analysis_gender = business[["gender_numeric","resilience_numeric"]].dropna()
female = analysis_gender.loc[analysis_gender["gender_numeric"]==0,"resilience_numeric"]
male = analysis_gender.loc[analysis_gender["gender_numeric"]==1,"resilience_numeric"]
u_statistic, p_value = mannwhitneyu(female, male, alternative="two-sided")
print("Mann-Whitney U:", u_statistic)
print("P-value:", p_value)
print("Female N:",len(female))
print("Male N:", len(male))
```

Since the p-value is much smaller that 0.05, we reject null hypothesis that the resilience distributions are the same across the two gender groups

There was a statistically significant difference in retail pharmacy business resilience by gender
